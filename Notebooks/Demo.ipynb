## 6. Alpha diversity (via scikit-bio)

```python
from skbio.diversity import alpha_diversity
import pandas as pd

matrix = pd.read_csv("Workspaces/DemoStudy/results/abundance_matrices/stage_04_final_assembly/abundance_matrix.csv", index_col=0)
shannon = alpha_diversity("shannon", matrix.values, ids=matrix.index.tolist())
print(shannon.sort_values(ascending=False))
```
